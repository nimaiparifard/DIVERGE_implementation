```python
from LLMReasoner.TAPE.peft_tape_finetuning.dataset_loader import load_dataset
from LLMReasoner.TAPE.peft_tape_finetuning.config import setup_finetuning_cfg

dataset_name = "citeseer"
cfg = setup_finetuning_cfg(dataset_name=dataset_name, llm_name="llama_3.2_1B", peft_type="lora")
dataset = load_dataset(cfg)
```

```python
def show_5_examples_for_Each_label(dataset, num_examples=5, max_text_len=400):
    """Show up to num_examples raw texts for each label in the dataset."""
    y = dataset.y.cpu().numpy() if hasattr(dataset.y, "cpu") else dataset.y
    label_names = dataset.label_name
    raw_texts = dataset.raw_texts

    for label_id, label_name in enumerate(label_names):
        indices = (y == label_id).nonzero()[0]
        print("=" * 60)
        print(f"Label: {label_name} (id={label_id}) — {len(indices)} nodes")
        print("=" * 60)
        for i, idx in enumerate(indices[:num_examples]):
            idx = int(idx)
            text = raw_texts[idx]
            snippet = text[:max_text_len] + ("..." if len(text) > max_text_len else "")
            print(f"\n  Example {i+1} (node {idx}):\n  {snippet}\n")
        if len(indices) > num_examples:
            print(f"  ... and {len(indices) - num_examples} more.\n")
        print()

show_5_examples_for_Each_label(dataset)
```
